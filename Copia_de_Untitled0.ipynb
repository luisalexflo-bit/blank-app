```python
import numpy as np

def sigmoide(x):
    return 1 / (1 + np.exp(-x))

def derivada_sigmoide(x):
    return x * (1 - x)

np.random.seed(42)

# 1. Entradas y el resultado exacto que queremos que aprenda a dar (objetivo: 1.0)
entradas = np.array([0.5, 1.2])
objetivo = np.array([[1.0]])

# 2. Inicializamos pesos y sesgos al azar
pesos_oculta = np.random.randn(2, 3)
sesgos_oculta = np.random.randn(3)
pesos_salida = np.random.randn(3, 1)
sesgo_salida = np.random.randn(1)

learning_rate = 0.5  # Qué tan rápido aprende la red
epocas = 10000       # Cuántas veces va a practicar

# 3. El ciclo de entrenamiento (¡aquí es donde aprende a la fuerza!)
for epoca in range(epocas):
    # Fase de predicción (Forward Pass)
    salida_oculta = sigmoide(np.dot(entradas, pesos_oculta) + sesgos_oculta)
    salida_final = sigmoide(np.dot(salida_oculta, pesos_salida) + sesgo_salida)

    # Medimos el error
    error = salida_final - objetivo

    # Fase de aprendizaje (Backpropagation)
    gradiente_salida = error * derivada_sigmoide(salida_final)

    error_oculta = np.dot(gradiente_salida, pesos_salida.T)
    gradiente_oculta = error_oculta * derivada_sigmoide(salida_oculta)

    # Actualizamos pesos y sesgos de ambas capas
    pesos_salida -= learning_rate * np.dot(salida_oculta.reshape(-1, 1), gradiente_salida.T)
    sesgo_salida -= learning_rate * gradiente_salida

    pesos_oculta -= learning_rate * np.dot(entradas.reshape(-1, 1), gradiente_oculta)
    sesgos_oculta -= learning_rate * gradiente_oculta

print("¡Entrenamiento terminado!")
print("Resultado final después de entrenar:", salida_final[0][0])
print("Valor objetivo deseado:", objetivo[0][0])
```

```python
import numpy as np

class RedNeuralPro:
    def __init__(self, input_size, hidden_size, output_size):
        np.random.seed(42)
        # Inicializamos pesos y sesgos de forma genérica
        self.pesos_oculta = np.random.randn(input_size, hidden_size)
        self.sesgos_oculta = np.random.randn(hidden_size)
        self.pesos_salida = np.random.randn(hidden_size, output_size)
        self.sesgo_salida = np.random.randn(output_size)

    def sigmoide(self, x):
        return 1 / (1 + np.exp(-x))

    def derivada_sigmoide(self, x):
        return x * (1 - x)

    def entrenar(self, X, y, learning_rate=0.5, epocas=20000):
        for epoca in range(epocas):
            for i in range(len(X)):
                entradas = X[i]
                objetivo = y[i]

                # 1. Forward Pass (Predicción)
                salida_oculta = self.sigmoide(np.dot(entradas, self.pesos_oculta) + self.sesgos_oculta)
                salida_final = self.sigmoide(np.dot(salida_oculta, self.pesos_salida) + self.sesgo_salida)

                # 2. Backpropagation (Aprendizaje)
                error = salida_final - objetivo
                gradiente_salida = np.atleast_2d(error * self.derivada_sigmoide(salida_final))

                error_oculta = np.dot(gradiente_salida, self.pesos_salida.T)
                gradiente_oculta = (error_oculta * self.derivada_sigmoide(salida_oculta)).reshape(-1)

                # 3. Actualización de pesos y sesgos
                self.pesos_salida -= learning_rate * np.dot(salida_oculta.reshape(-1, 1), gradiente_salida)
                self.sesgo_salida -= learning_rate * gradiente_salida.reshape(-1)

                self.pesos_oculta -= learning_rate * np.dot(entradas.reshape(-1, 1), gradiente_oculta.reshape(1, -1))
                self.sesgos_oculta -= learning_rate * gradiente_oculta

    def predecir(self, X):
        predicciones = []
        for i in range(len(X)):
            entradas = X[i]
            salida_oculta = self.sigmoide(np.dot(entradas, self.pesos_oculta) + self.sesgos_oculta)
            salida_final = self.sigmoide(np.dot(salida_oculta, self.pesos_salida) + self.sesgo_salida)
            predicciones.append(salida_final[0])
        return predicciones

# --- Probando nuestra clase con el reto XOR ---
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# Creamos nuestra propia instancia de la red (2 entradas, 4 neuronas ocultas, 1 salida)
mi_red = RedNeuralPro(input_size=2, hidden_size=4, output_size=1)

print("Entrenando la red orientada a objetos...")
mi_red.entrenar(X, y, learning_rate=0.5, epocas=20000)

print("\n¡Listo! Evaluando predicciones finales:")
resultados = mi_red.predecir(X)
for i in range(len(X)):
    print(f"Entrada: {X[i]} -> Predicción: {resultados[i]:.4f} (Esperado: {y[i][0]})")
```
